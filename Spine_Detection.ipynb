*Lets check using a statistical test whether the distribution of sum of (pelvic_tilt, sacral_slope) and pelvic_incidence are same*
*Before starting a statistical test, we should check whether our distribution is normal, If normal we will use a parametric test else we will use a non-parametric test.*

*Checking the normality of pelvic_incidence*

*There are 3 commonly used statistical test for normality:*

- Shapiro-Wilk Test
- D’Agostino’s K^2 Test
- Anderson-Darling Test

*In our context we will be using Shapiro-Wilk Test as it is suitable for smaller samples of data*

# add reference

**Null Hypothesis (H<sub>0</sub>) =  Sample was drawn from a Gaussian distribution**

**Alternate Hypothesis(H<sub>a</sub>) = Sample was  not drawn from a Gaussian distribution**

- **p <= alpha: reject H0, not normal.**
- **p > alpha: fail to reject H0, normal.**

```python
# Shapiro-Wilk Test
from numpy.random import seed
from numpy.random import randn
from scipy.stats import shapiro
data = list(df.pelvic_incidence)
# normality test
stat, p = shapiro(data)
print('Statistics=%.3f, p=%.3f' % (stat, p))
# interpret
alpha = 0.05
if p > alpha:
    print('Sample looks Gaussian (fail to reject H0)')
else:
    print('Sample does not look Gaussian (reject H0)')
```

*Lets now check a Q-Q plot*

***Q-Q Plots*** *(Quantile-Quantile plots) are plots of two quantiles against each other. A quantile is a fraction where certain values fall below that quantile. For example, the median is a quantile where 50% of the data fall below that point and 50% lie above it. The purpose of Q Q plots is to find out if two sets of data come from the same distribution. A 45 degree angle is plotted on the Q-Q plot; if the two data sets come from a common distribution, the points will fall on that reference line.*

```python
# QQ Plot
from statsmodels.graphics.gofplots import qqplot
data = df.pelvic_incidence
# q-q plot
qqplot(data, line='s')
plt.title("Q-Q plot for the Pelvic Incednce")
plt.show()
```

*From the Q-Q plot we can see that there are not much points on the line. So , If all the points were on the line then we could have said that it is normally distributed.*

*From the histogram and Q-Q plots we got to know that pelvic_incidence is not normal.*

*Now that we know that the pelvic_incidence column is not normal, we can use non-parametric test.*

*Statistical tests that we will be using are Mann-Whitney U Test and Wilcoxon Signed-Rank Test*

*The* ***Mann-Whitney U test*** *is a nonparametric statistical significance test for determining whether two independent samples were drawn from a population with the same distribution.*

**Null Hypothesis (H<sub>0</sub>) =  There is no difference between the distributions of the data samples**

**Alternate Hypothesis(H<sub>a</sub>) = There is a difference between the distributions of the data samples**


*Fail to Reject H<sub>0</sub>: Sample distributions are equal.*

*Reject H<sub>0</sub>: Sample distributions are not equal.*

```python
# Mann-Whitney U test
from scipy.stats import mannwhitneyu
data1 = list(df.pelvic_incidence)
data2 = [sum(n) for n in zip(*[list(df.pelvic_tilt), list(df.sacral_slope)])] # to add corresponding elements of two lists
# compare samples
stat, p = mannwhitneyu(data1, data2)
print('Statistics=%.3f, p=%.3f' % (stat, p))
# interpret
alpha = 0.05
if p > alpha:
    print('Same distribution (fail to reject H0)')
else:
    print('Different distribution (reject H0)')
```

*Mann-Whitney U test is telling us that these are same distributions, let try Wilcoxon Signed-Rank Test and then we will try to make some inference out of it.*

*In some cases, the data samples may be paired.*

*There are many reasons why this may be the case, for example, the samples are related or matched in some way or represent two measurements of the same technique. More specifically, each sample is independent, but comes from the same population.*

*Examples of paired samples in machine learning might be the same algorithm evaluated on different datasets or different algorithms evaluated on exactly the same training and test data.*

" ***The Wilcoxon signed ranks test*** *is a nonparametric statistical procedure for comparing two samples that are paired, or related. The parametric equivalent to the Wilcoxon signed ranks test goes by names such as the Student’s t-test, t-test for matched pairs, t-test for paired samples, or t-test for dependent samples."*


**Null Hypothesis (H<sub>0</sub>) =  The two samples have the same distribution.**

**Alternate Hypothesis(H<sub>a</sub>) = The two samples don't have the same distribution**

- Fail to Reject H<sub>0</sub>: Sample distributions are equal.
- Reject H<sub>0</sub>: Sample distributions are not equal.

```python
# Wilcoxon signed-rank test
from scipy.stats import wilcoxon
# seed the random number generator
seed(1)
data1 = list(df.pelvic_incidence)
data2 = [sum(n) for n in zip(*[list(df.pelvic_tilt), list(df.sacral_slope)])] # to add corresponding elements of two lists
# compare samples
stat, p = wilcoxon(data1, data2)
print('Statistics=%.3f, p=%.3f' % (stat, p))
# interpret
alpha = 0.05
if p > alpha:
    print('Same distribution (fail to reject H0)')
else:
    print('Different distribution (reject H0)')
```

*But before jumping over, I will try to explain the evaluation matrices for our classification problem in context with this problem.*

- **True Positive:** *The cases in which we predicted "Abnormal" and the actual output was also "Abnormal".*
- **True Negatives:** *The cases in which we predicted "Normal" and the actual output was also "Normal".*
- **False Positives:** *The cases in which we predicted "Abnormal" but the actual output was "Normal".*
- **False Negatives:** *The cases in which we predicted "Normal" but the actual output was "Abnormal".*

<h3>Evaluation Metrics</h3>

- **Accuracy:** *Ratio of the cases in which we predicted "Abnormal" over the total number of Cases*
- **Precision:** *Out of all those cases we predicted "Abnormal", how many of them are actually "Abnormal"*
- **Recall:** *Out of all those cases where it was actually "Abnormal", how many of them we predicted truly "Abnormal"*
- **F1 Score:** *It is a function of Precision and Recall. It is needed when you want to seek a balance between Precision and Recall*

**So when to use Precision and when to use recall ?**

*In some contexts you mostly care about precision, and in other contexts you really care about recall. Lets take an example, If you regularly spend your weekends on watching Netflix, You might have come across a Kids section. And suppose you built a classifier to classify videos on whether it is good for kids or not, you would probably prefer a classifier that rejects many good videos (low recall) but keeps only safe ones (high precision), rather than a classifier that has a much higher recall but lets a few really bad videos show up in your product.*

*On the other hand, suppose you train a classifier to detect shoplifters on surveillance images: it is probably fine if your classifier has only 30% precision as long as it has 99% recall (sure, the security guards will get a few false alerts, but almost all shoplifters will get caught).*

*In our context, It's okay if we have very low precision as long as we have very good recall. We will get some false results but for them we can do the tests and scans again but we will be able detect all the Abnormalities.*
